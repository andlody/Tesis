```python
docs = (df["CHUNK_PRODUCTO"]+'. '+df["CHUNK_FICHA"]+'. '+df["CHUNK_DESCRIPCION"]+'. '+df["CHUNK_CARACTERISTICAS"]+'. '+df["CHUNK_OBSERVACIONES"]+'. '+df["CHUNK_RECOMENDACIONES"]).tolist()
docs
```

```python
from sentence_transformers import SentenceTransformer

# Cargar modelo localmente
model = SentenceTransformer(MAIN_ROUTE+"/modelos/all-MiniLM-L6-v2")
embeddings = model.encode(docs, show_progress_bar=True, convert_to_numpy=True)
```

```python
import numpy as np
# ---------- 3) Normalizar similaridad coseno ----------
def normalize(x: np.ndarray):
    norms = np.linalg.norm(x, axis=1, keepdims=True)
    return x / (norms + 1e-10)

embeddings = normalize(embeddings)
```

```python

np.save(MAIN_ROUTE+"/embeddings/CHUNK_TOTAL.npy", embeddings)
```

```python
import faiss
from typing import List

# ---------- 1) Datos de ejemplo ----------
doc_ids = [f"doc_{i}" for i in range(len(docs))]


# ---------- 4) Crear índice FAISS (IndexFlatIP para coseno con vectores normalizados) ----------
d = embeddings.shape[1]
index = faiss.IndexFlatIP(d)        # producto interno
index.add(embeddings)              # añade vectores

id_to_doc = dict(zip(range(len(doc_ids)), docs))


# ---------- 5) Función de búsqueda ----------
def retrieve(query: str, k: int = 3) -> List[str]:
    q_emb = model.encode([query], convert_to_numpy=True)
    q_emb = normalize(q_emb)
    D, I = index.search(q_emb, k)  # D: similitud, I: índices
    results = []
    for idx in I[0]:
        if idx == -1: continue
        results.append(id_to_doc[idx])
    return results
```

```python
# ---------- 6) Ejemplo RAG: consulta + prompt para SLM ----------
query = "casa ducha rompio"
retrieved = retrieve(query, k=3)
context = "\n\n".join(retrieved)

# Construir prompt para SLM 
prompt = f"""
Contexto (documentos relevantes):
{context}

Pregunta: {query}

Usando solo la información del contexto, responde de forma concisa y paso a paso. Si la información no está, di que no hay suficientes detalles.
"""

print(prompt)
```

```python
import spacy
from spacy.lang.es.stop_words import STOP_WORDS

VERBOS_RUIDO = ["necesitar", "querer", "buscar", "requerir", "comprar", "utilizar",'cosito','adecuar','cualquier','funcionar','usar','']

nlp = spacy.load("es_core_news_sm")

def palabras_relevantes(texto):
    doc = nlp(texto)

    palabras = []
    for token in doc:
        if (token.pos_ in ["NOUN", "VERB", "PROPN"]) and token.text.lower() not in STOP_WORDS:
            lemma = token.lemma_.lower()
            if lemma not in VERBOS_RUIDO:
                palabras.append(lemma)

    return " ".join(palabras)

# Ejemplo
oracion = "dentro de mi casa tengo una ducha, necesito el cosito para la ducha que se rompio."
resultado = palabras_relevantes(oracion)
print(resultado)
```

```python
dp = pd.read_csv("preguntas-respuestas-v1.1.csv",delimiter=";", encoding='utf-8')

dp
```

```python
dp['NER'] = dp['Pregunta'].apply(palabras_relevantes)
dp
```
