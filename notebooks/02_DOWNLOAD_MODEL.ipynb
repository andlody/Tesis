```python
from sentence_transformers import SentenceTransformer
from transformers import AutoTokenizer, AutoModelForCausalLM
```

```python
# Descargar y guardar el modelo de embeddings localmente
embedder = SentenceTransformer("all-MiniLM-L6-v2")
embedder.save(MAIN_ROUTE+"/modelos/all-MiniLM-L6-v2")
logging.info("Se descarga el modelo (all-MiniLM-L6-v2)")
```

```python
# Descargar y guardar el modelo de embeddings localmente
embedder = SentenceTransformer("all-MiniLM-L12-v2")
embedder.save(MAIN_ROUTE+"/modelos/all-MiniLM-L12-v2")
logging.info("Se descarga el modelo (all-MiniLM-L12-v2)")
```

```python
logging.info("[Inicio] Se descarga el modelo (Phi-4-mini-instruct)")
# Descargar y guardar el tokenizer y modelo de lenguaje
modelo_id = "microsoft/Phi-4-mini-instruct"
tokenizer = AutoTokenizer.from_pretrained(modelo_id)
tokenizer.save_pretrained(MAIN_ROUTE+"/modelos/Phi-4-mini-instruct")

modelo = AutoModelForCausalLM.from_pretrained(modelo_id)
modelo.save_pretrained(MAIN_ROUTE+"/modelos/Phi-4-mini-instruct")
logging.info("[Fin] Se descarga el modelo (Phi-4-mini-instruct)")
```
